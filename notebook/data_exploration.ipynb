```python

# ============================================================
# PREDICTX AI - EXPLORATORY DATA ANALYSIS
# NASA C-MAPSS FD001
# ============================================================

# 1. Basic dataset validation
print("=" * 60)
print("PREDICTX AI - DATASET OVERVIEW")
print("=" * 60)

assert not train_df.empty, "Training dataset is empty!"
assert train_df.shape[1] == 26, "Unexpected number of columns!"

print("Training records:", len(train_df))
print("Total engines:", train_df["unit_id"].nunique())
print("Total features:", len(columns))

# 2. Data quality analysis
print("\n--- DATA QUALITY ---")

print("Missing values:", train_df.isnull().sum().sum())
print("Duplicate rows:", train_df.duplicated().sum())

print("\n--- DATA TYPES ---")
print(train_df.dtypes)

# 3. Engine lifecycle analysis
engine_life = (
    train_df.groupby("unit_id")["cycle"]
    .max()
    .rename("total_cycles")
)

print("\n--- ENGINE LIFECYCLE STATISTICS ---")
print(engine_life.describe())

# 4. Sensor statistical analysis
sensor_cols = [col for col in train_df.columns
               if col.startswith("sensor_")]

sensor_summary = train_df[sensor_cols].describe().T

sensor_summary["variance"] = train_df[sensor_cols].var()
sensor_summary["unique_values"] = train_df[sensor_cols].nunique()

print("\n--- SENSOR SUMMARY ---")
display(sensor_summary)

# 5. Engine lifecycle distribution
plt.figure(figsize=(10, 5))

sns.histplot(
    engine_life,
    bins=20,
    kde=True
)

plt.title("Distribution of Engine Lifecycles")
plt.xlabel("Total Operational Cycles")
plt.ylabel("Number of Engines")
plt.tight_layout()
plt.show()

# 6. Sensor variance analysis
plt.figure(figsize=(12, 6))

sensor_summary["variance"].sort_values().plot(
    kind="bar"
)

plt.title("Variance Across Sensor Measurements")
plt.xlabel("Sensor")
plt.ylabel("Variance")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 7. Sensor degradation trend analysis
# Median sensor reading across engines at each cycle

fig, axes = plt.subplots(7, 3, figsize=(18, 25))
axes = axes.flatten()

for i, sensor in enumerate(sensor_cols):

    cycle_median = (
        train_df.groupby("cycle")[sensor]
        .median()
    )

    axes[i].plot(
        cycle_median.index,
        cycle_median.values
    )

    axes[i].set_title(f"{sensor} vs Cycle")
    axes[i].set_xlabel("Operational Cycle")
    axes[i].set_ylabel("Median Reading")
    axes[i].grid(alpha=0.3)

plt.suptitle(
    "Sensor Behaviour Across Engine Lifecycles",
    fontsize=16,
    y=1.002
)

plt.tight_layout()
plt.show()

print("\nEDA completed successfully!")
```
